```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / "data/processed/train_semantic.csv").exists():
    ROOT = ROOT.parent

real = pd.read_csv(
    ROOT / "data/processed/train_semantic.csv"
)

synthetic = pd.read_csv(
    ROOT / "data/synthetic/tvae_50ep.csv"
)

print("Real shape:", real.shape)
print("Synthetic shape:", synthetic.shape)

print("\nReadmission rates (%)")
for name, df in [("Real", real), ("TVAE", synthetic)]:
    rate = (df["readmitted_30"].astype(str) == "1").mean() * 100
    print(f"{name}: {rate:.2f}%")

print("\nPrior hospital utilization (mean)")
cols = [
    "number_outpatient",
    "number_emergency",
    "number_inpatient"
]

print(pd.DataFrame({
    "Real": real[cols].mean(),
    "TVAE": synthetic[cols].mean()
}).round(3))

print("\nCategory coverage")
for col in ["race", "diag_3"]:
    print(
        col,
        "Real:", real[col].nunique(),
        "TVAE:", synthetic[col].nunique()
    )
```

```python
synthetic_300 = pd.read_csv(
    ROOT / "data/synthetic/tvae_300ep.csv"
)

print("Prior hospital utilization (mean)")
cols = [
    "number_outpatient",
    "number_emergency",
    "number_inpatient"
]

print(pd.DataFrame({
    "Real": real[cols].mean(),
    "TVAE 50": synthetic[cols].mean(),
    "TVAE 300": synthetic_300[cols].mean()
}).round(3))

print("\nCategory coverage")
for col in ["race", "diag_3"]:
    print(
        col,
        "Real:", real[col].nunique(),
        "TVAE 50:", synthetic[col].nunique(),
        "TVAE 300:", synthetic_300[col].nunique()
    )
```
